```python
import cv2
import os

letakVidio = "VideoDelay.mp4"
file_counter = 1
output_folder = "frames"
cap = cv2.VideoCapture(letakVidio)
ret = True
while ret == True:
    ret, frame = cap.read()
    
    if ret:
        filename = os.path.join(output_folder, f'frame_{file_counter:04d}.png')
        cv2.imwrite(filename, frame)
        file_counter += 1
```
